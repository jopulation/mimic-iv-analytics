```python
import pandas as pd
import os

# 데이터 폴더 경로 (본인 환경에 맞게 수정)
base_path = '../data/'
modules = ['hosp', 'icu']

print("=== MIMIC-IV Demo Data Summary ===\n")

for module in modules:
    module_path = os.path.join(base_path, module)
    if not os.path.exists(module_path):
        continue
        
    print(f"📂 Module: {module.upper()}")
    print("-" * 60)
    
    # 폴더 내 모든 파일 탐색
    files = sorted([f for f in os.listdir(module_path) if f.endswith('.csv.gz')])
    
    for file in files:
        file_path = os.path.join(module_path, file)
        try:
            # 압축된 CSV 읽기 (헤더만 읽거나 전체 읽어서 카운트)
            # 데모 버전은 용량이 작아서 다 읽어도 됩니다.
            df = pd.read_csv(file_path, compression='gzip')
            
            print(f"📄 {file:<25} | 행(Row): {len(df):>6,} | 열(Col): {len(df.columns)}")
            # 필요하면 컬럼명도 출력 가능
            # print(f"   Columns: {list(df.columns)}") 
            
        except Exception as e:
            print(f"❌ Error reading {file}: {e}")
    print("\n")
```
